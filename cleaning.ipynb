# CHECKING MISSING VALUES

```python
# Check missing values
df.isnull().sum()
```

### HANDLING MISSING VALUES

```python
#Convert the negative age value into NaN
df.loc[df["Age"] == -5, "Age"] = np.nan
print(df['Age'])
    
```

```python
#Replace the Null value using mean in Age column.
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
#Replace the Null value using mode in Department column.
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
```

```python
#Replave the Null value using median in Salary column.
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
```

```python
#Replace the Null value using mean in Experience column.
df['Experience'] = df['Experience'].fillna(df['Experience'].mean())
```

```python
#Check the Null values 
df.isnull().sum()
```

# CHECKING DUPLICATE VALUES

```python
#Checking Duplicates
df.duplicated().sum()
```

### HANDLING DUPLICATE VALUES

```python
df.duplicated(
    subset=["Name", "Age", "Department", "Salary", "Experience", "City", "Joining_Year"]
).sum()
```

```python
print(df[df.duplicated(
    subset=["Name", "Age", "Department", "Salary", "Experience", "City", "Joining_Year"],
    keep=False
)])
```

```python
df = df.drop_duplicates(
    subset=["Name", "Age", "Department", "Salary",
            "Experience", "City", "Joining_Year"]
)
```

# CHECKING DATATYPES

```python
#Checking Datatypes
df.dtypes
```

### HANDLING DATATYPES

```python
#Converting Age column datatype into int.
df["Age"] = df["Age"].astype(int)
```

```python
#Converting Experience column datatype into int.
df["Experience"] = df["Experience"].astype(int)
```

```python
#Checking Data types.
df.dtypes
```

## HANDLING INCONSISTENT DATA

```python
#Handle inconsistent data.
df["City"] = df["City"].str.lower().str.strip()

df['City'] = df['City'].map({
    'chennai' : 'Chennai',
    'coimbatore': 'Coimbatore'
})
```

```python
df["Department"] = df["Department"].str.lower().str.strip()

df['Department'] = df['Department'].map({
    'data':'Data',
    'hr':'HR',
    'it':'IT'
})
```

## CHECKING OUTLIERS

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Salary"] < lower) |
    (df["Salary"] > upper)
]

print(outliers)
```
